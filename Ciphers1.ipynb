# Option 1: Shift Cipher

## Shift Encrypt Function

```python
# function to encrypt plain text into shift chipher text
def eshift(PT, S):
    result = ""

    # shifting text
    for i in range(len(PT)):  # for i in the length of the plaintext to convert
        char = PT[i]  # variable character equals to the length of the plaintext character

        # encrypt plaintext characters
        if char.isupper():  # if uppercase is entered
            result += chr((ord(char) - 65 + S) % 26 + 65)
            # encrypts a letter using Caesar cipher with shift S
        elif char.islower():  # if lowercase is entered
            result += chr((ord(char) - 97 + S) % 26 + 97)
            # encrypts a lowercase letter using Caesar cipher with shift S
        else:
            result += char  # keeps spaces and punctuation unchanged
    return result  # return above
```

## Shift Decrypt Function

```python
# function to decrypt shift cipher text into plain text
def dshift(CT, S):
    result = ""

    # shifting text
    for i in range(len(CT)):  # for i in the length of the ciphertext to convert
        char = CT[i]  # variable character equals to the length of the ciphertext character

        # decrypt ciphertext characters
        if char.isupper():  # if uppercase is entered
            result += chr((ord(char) - 65 - S) % 26 + 65)
            # decrypts a letter using Caesar cipher with shift S
        elif char.islower():  # if lowercase is entered
            result += chr((ord(char) - 97 - S) % 26 + 97)
            # decrypts a lowercase letter using Caesar cipher with shift S
        else:
            result += char  # keeps spaces and punctuation unchanged
    return result  # return above
```

# Option 2: Symmetric-key

## Symmetric-key Encryption Function

```python
# function to encrypt plain text into shift chipher text
def esymkey(PT, K = None):
    # generate a new key if none is provided
    key = K if K else Fernet.generate_key()
    # create a fernet object using the key
    fernet = Fernet(key)
    # encrypt the plaintext and return both ciphertext and key
    Enkey = fernet.encrypt(PT.encode())
    return Enkey, key
```

## Symmetric-key Decryption Function

```python
# function to decrypt Symmetric-key cipher text into plain text
def dsymkey(CT, K):
    # create a fernet object using the provided key
    fernet = Fernet(K)
    # decrypt the ciphertext and decode to string
    PT = fernet.decrypt(CT).decode()
    return PT
```

# User Input

```python
from cryptography.fernet import Fernet

# start an infinite loop to prompt user until valid input is given
while True:
    # ask user to choose encryption or decryption
    crypt = input("Enter E for Encryption or D for Decryption: ")

    if crypt == "E":
        # prompt for encryption method
        method = input("Enter Encryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA\n\n")
        
        if method == "1":    
            # get plaintext and shift value
            PT = input("\nEnter Plaintext Passage: ")
            S = int(input("\nShift Value: "))
            CT = eshift(PT, S)  # encrypt using shift cipher
            print("\nCiphertext: ", CT)
            break  # exit loop after encryption
           
        if method == "2":    
            # get plaintext and encrypt using symmetric key
            PT = input("\nEnter Plaintext Passage: ")
            CT, K = esymkey(PT)
            print("\nEncryption Key: ", K.decode())
            print("\nCiphertext: ", CT)
            break  # exit loop after encryption

    elif crypt == "D":
        # prompt for decryption method
        method = input("Enter Decryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA \n\n")
        
        if method == "1": 
            # get ciphertext and shift value
            CT = input("\nEnter Cipher Passage: ")
            S = int(input("\nShift Value: "))
            PT = dshift(CT, S)  # decrypt using shift cipher
            print("\nPlaintext: ", PT)
            break  # exit loop after decryption

        if method == "2": 
            # get ciphertext and key for symmetric decryption
            CTI = input("\nEnter Ciphertext (base64 ONLY): ")
            KI = input("Enter Key (base64 ONLY): ")
        
            # convert inputs to bytes
            CT = CTI.encode()
            K = KI.encode()
        
            PT = dsymkey(CT, K)  # decrypt using symmetric key
            print("\nDecryption Key:", K.decode())
            print("\nPlaintext:", PT)
            break  # exit loop after decryption
            
    else:
        # handle invalid input
        print("Invalid Option. Please Enter E or D.\n")
        continue  # repeat loop
```
