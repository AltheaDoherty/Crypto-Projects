```python
#Function to retry entry again
def retry():
    while True:
        again = input("New Game? Enter Y for Yes or Another Key to End: ")
        if again == "Y":
            print("Game Loading....")
            continue
        else:
            print("Game End...")
            break
```

```python
#importing libraries needed for GUI
import tkinter as tk

#creating a pop out window for program
window = tk.Tk()
window.geometry("850x450")  #window size
window.title("Cryptogram Puzzle")  #window title

#creating health bar
health = tk.IntVar(value=100)  #health bar status variable
hLabel = tk.Label(window, text="Health: 100")  #health bar label text
hLabel.pack(pady=10)  #spacing for health label

#creating label on top of where quote will sit
eLabel = tk.Label(window, text="Quote")  #label for encrypted quote
eLabel.pack(pady=10)

#creating label on top of where entry window is
pLabel = tk.Label(window, text="Progress")  #label for user guess area
pLabel.pack(pady=10)

#size of the entry window frame
pFrame = tk.Frame(window)  #frame to hold encrypted letters + entry boxes
pFrame.pack(pady=5)

#size of the guess window frame
gFrame = tk.Frame(window)  #frame to hold guess history
gFrame.pack(pady=5)

#puzzle begin
encrypted ="Qrwklqj'v shupdqhqw."  #encrypted text shown to player
correct = "NOTHING'S PERMANENT."    #correct decrypted solution
corAns = [ch.upper() for ch in correct if ch.isalpha()]

#empty list to hold entry widgets for each letter
entries = []

#loop through encrypted text and create labels + entry boxes
for i, ch in enumerate(encrypted):
    label = tk.Label(pFrame, text=ch, font=("Times",12))  #show encrypted letter
    label.grid(row=0, column=i, padx=3)

    if ch.isalpha():  #only create entry box if character is a letter
        entry = tk.Entry(pFrame, width=3, font=("Terminal",12), justify="center")
        entry.grid(row=1, column=i, padx=3)
        entries.append(entry)  #store entry for later checking
    else:
        tk.Label(pFrame, text=" ").grid(row=1, column=i)  #placeholder for non-letters

#function to update health bar
def hLvl(amount):
    #reduce health but never below 0
    hupdate = max(0, health.get() - amount)
    health.set(hupdate)
    hLabel.config(text=f"Health: {hupdate}")  #update label text

#function to show correctness of guesses
def subGuess():
    #get all guesses from entry boxes, convert to uppercase
    guessl = (e.get().upper() for e in entries)
    wrong = False  #track if any guess is wrong

    #first pass: check each guess and color boxes
    for i, guess in enumerate(guessl):
        if guess == correct[i]:  #correct letter
            entries[i].config(bg="limegreen")
        else:  #incorrect letter
            entries[i].config(bg="red")
            wrong = True

    #if any wrong guesses, reduce health
    if wrong:
        hLvl(10)

    #second pass: auto-fill matching letters in all positions
    guessl = (e.get().upper() for e in entries)  #regenerate generator
    for i, guess in enumerate(guessl):
        if guess and i < len(corAns) and guess == corAns[i]: #making it possible for guesses even when full length of text has not been entered
            for x, actual in enumerate(corAns):
                if actual == guess:
                    entries[i].delete(0, tk.END)
                    entries[i].insert(0, guess)
                    entries[i].config(bg="limegreen")

    #history of incorrect/correct guesses
    row = tk.Frame(gFrame)
    row.pack()

    for i, e in enumerate(entries):
        tk.Label(row,text=e.get().upper(),width=3,font=("Terminal",14),justify="center",bg=e.cget("bg")).grid(row=0, column=i, padx=3, pady=3)

#creating submit button to turn in guesses for puzzle
subButton = tk.Button(window, text="Submit Entry", command=subGuess)
subButton.pack(pady=10)

#calling main loop to run program
window.mainloop()
```
