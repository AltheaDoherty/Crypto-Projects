```python
from tkinter import *
from tkinter.ttk import *

window = Tk()
window.geometry("750x550")
window.title("Cryptogram Puzzle")

eLabel = Label(window, text="Quote")
eLabel.pack(pady=30)

pLabel = Label(window, text="Progress")
pLabel.pack(pady=30)

pFrame = Frame(window)
pFrame.pack(pady=10)

encrypted = "KRKT UO BAI GOJD"
correct   = "THIS IS THE TEXT"

entries = []

for i, ch in enumerate(encrypted):
    label = Label(pFrame, text=ch, font=("Times",18))
    label.grid(row=0, column=i, padx=3)

    if ch.isalpha():
        entry = Entry(pFrame, width=3, font=("Terminal",14), justify="center")
        entry.grid(row=1, column=i, padx=3)
        entries.append((i,entry))
    else:
        Label(pFrame, text=" ").grid(row=1,column=i)

def subGuess():
    for i, entry in entries:
        guess = entry.get().upper()
        actual = correct[i]

        if guess == actual:
            entry.config(background="cornflowerblue")
        else:
            entry.config(background="magenta")

subButton = Button(window, text="Submit Entry", command=subGuess)
subButton.pack(pady=10)

window.mainloop()
```
