# Option 1: Shift Cipher

## Shift Encrypt Function

```python
def retry():
    while True:
        again = input("\nRetry Entry? Enter Y for Yes or Another to End: ")
        if again == "Y":
            print("Retrying Encrytion/Decription Program.")
            continue
        else:
            print("Encrytion/Decription Program End.")
            break
```

```python
# Function to encrypt plaintext into shift cipher (Caesar cipher) text
def eshift(PT, S):
    result = ""

    # Iterate through each character in the plaintext
    for i in range(len(PT)):
        char = PT[i]

        # Encrypt uppercase letters
        if char.isupper():
            # Shift character forward by S positions in the alphabet
            result += chr((ord(char) - 65 + S) % 26 + 65)

        # Encrypt lowercase letters
        elif char.islower():
            # Shift character forward by S positions in the alphabet
            result += chr((ord(char) - 97 + S) % 26 + 97)

        # Leave non-alphabetic characters unchanged (e.g., spaces, punctuation)
        else:
            result += char

    # Return the encrypted ciphertext
    return result
```

## Shift Decrypt Function

```python
# Function to decrypt shift cipher (Caesar cipher) text into plaintext
def dshift(CT, S):
    result = ""

    # Iterate through each character in the ciphertext
    for i in range(len(CT)):
        char = CT[i]

        # Decrypt uppercase letters
        if char.isupper():
            # Shift character backward by S positions in the alphabet
            result += chr((ord(char) - 65 - S) % 26 + 65)

        # Decrypt lowercase letters
        elif char.islower():
            # Shift character backward by S positions in the alphabet
            result += chr((ord(char) - 97 - S) % 26 + 97)

        # Leave non-alphabetic characters unchanged (e.g., spaces, punctuation)
        else:
            result += char

    # Return the decrypted plaintext
    return result
```

# Option 2: Symmetric-key

## Symmetric-key Encryption Function

```python
# Function to encrypt plaintext into ciphertext using symmetric-key encryption (Fernet)
def esymkey(PT, K=None):
    # If no key is provided, generate a new Fernet key (base64-encoded)
    key = K if K else Fernet.generate_key()
    # Create a Fernet object using the key
    fernet = Fernet(key)
    # Encrypt the plaintext (converted to bytes) and return the ciphertext
    Enkey = fernet.encrypt(PT.encode())
    # Return both the encrypted ciphertext and the key used
    return Enkey, key
```

## Symmetric-key Decryption Function

```python
# Function to decrypt symmetric-key ciphertext into plaintext using Fernet
def dsymkey(CT, K):
    # Create a Fernet object using the provided key (K must be in base64 format)
    fernet = Fernet(K)
    # Decrypt the ciphertext (CT) and decode the result from bytes to string
    PT = fernet.decrypt(CT).decode()
    # Return the decrypted plaintext
    return PT
```

```python
# Function to generate a new RSA public-private key pair
def genKey():
    # Generate a 512-bit RSA key pair (public and private keys)
    publicKey, privateKey = rsa.newkeys(512)
    # Return both keys for use in encryption and decryption
    return publicKey, privateKey
```

```python
# Function to encrypt a plaintext message using RSA and a public key
def easym(PT: str, publicKey):
    # Convert the plaintext string to bytes and encrypt it using the RSA public key
    CT = rsa.encrypt(PT.encode(), publicKey)
    # Return the encrypted ciphertext as bytes (can be base64-encoded later)
    return CT
```

## Asymmetric-key Decryption

```python
# Function to decrypt a base64-encoded RSA ciphertext using a private key
def dasym(CT: str, privateKey):
    # Remove any leading/trailing whitespace from the ciphertext
    CT = CT.strip()

    # Ensure the base64 string has correct padding (length must be a multiple of 4)
    padding = len(CT) % 4
    if padding:
        CT += "=" * (4 - padding)

    # Decode the base64 string into raw encrypted bytes
    CTbytes = base64.b64decode(CT)

    # Decrypt the bytes using the provided RSA private key and decode to plaintext string
    PT = rsa.decrypt(CTbytes, privateKey).decode()

    # Return the decrypted plaintext
    return PT
```

```python
from cryptography.fernet import Fernet
import rsa
import base64

# Start an infinite loop to prompt user until valid input is given
while True:
    # Ask user to choose encryption or decryption
    crypt = input("\n Enter E for Encryption, D for Decryption, or 0 to End: ")

    if crypt == "E":
        # Prompt for encryption method
        method = input("Enter Encryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA(base64)\n \n R: Return to Main Menu\n")
        
        if method == "1":    
            # Shift Cipher Encryption
            PT = input("\nEnter Plaintext Passage: ")
            S = int(input("\nShift Value: "))
            CT = eshift(PT, S)  # Encrypt using shift cipher
            print("\nCiphertext: ", CT)
            break  # Exit loop after encryption
           
        elif method == "2":    
            # Symmetric Key Encryption using Fernet (base64)
            PT = input("\nEnter Plaintext Passage: ")
            CT, K = esymkey(PT)  # Encrypt and generate key
            print("\nEncryption Key: ", K.decode())  # Show key
            print("\nCiphertext: ", CT)  # Show ciphertext
            break  # Exit loop after encryption

        elif method == "3":
            # RSA Encryption
            PT = input("Enter Plaintext Passage: ")
            publicKey, privateKey = genKey()  # Generate RSA key pair
            CT = easym(PT, publicKey)  # Encrypt plaintext with public key

            # Save keys for use in decryption
            savedPubKey = publicKey
            savedPrivKey = privateKey
            
            # Display keys and ciphertext
            print("\n", savedPubKey)
            print("\n", savedPrivKey)
            print("\nCiphertext(base64): ", CT)
            break  # Exit loop after encryption
        
        elif method == "R":
            continue  # Return to main menu
             
        else:
            # Handle invalid encryption method input
            method = input("\nInvalid Option. Enter Encryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA\n \n R: Return to Main Menu\n")
            
    elif crypt == "D":
        # Prompt for decryption method
        method = input("Enter Decryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA(base64) \n \n R: Return to Main Menu\n")
        
        if method == "1": 
            # Shift Cipher Decryption
            CT = input("\nEnter Cipher Passage: ")
            S = int(input("\nShift Value: "))
            PT = dshift(CT, S)  # Decrypt using shift cipher
            print("\nPlaintext: ", PT)
            break  # Exit loop after decryption

        if method == "2": 
            # Symmetric Key Decryption using Fernet (base64)
            CTI = input("\nEnter Ciphertext (base64): ")
            KI = input("Enter Key (base64): ")
        
            # Convert inputs to bytes
            CT = CTI.encode()
            K = KI.encode()
        
            PT = dsymkey(CT, K)  # Decrypt using symmetric key
            print("\nDecryption Key:", K.decode())
            print("\nPlaintext:", PT)
            break  # Exit loop after decryption

        elif method == "3":
            # RSA Decryption
            CT = input("Enter Ciphertext (base64): ")

            # Check if private key is available from previous encryption
            if "savedPrivKey" in globals():
                try:
                    PT = dasym(CT, savedPrivKey)  # Attempt decryption
                    print("Decrypted plaintext:", PT)
                    print("Private Key: ", savedPrivKey)
                except rsa.DecryptionError:
                    # Handle decryption failure due to key mismatch
                    print("Decryption Failed. Private Key is not a Valid Match.")
            else:
                # No key available to decrypt
                print("Unable to Decrypt.")   
                break  # Exit loop after decryption
                
        elif method == "R":
            continue  # Return to main menu
             
        else:
            # Handle invalid decryption method input
            method = input("\nInvalid Option. Enter Decryption Method 1, 2, or 3.\n 1: Shift Cipher \n 2: Symmetric-key(base64) \n 3: RSA\n \n R: Return to Main Menu\n")

    elif crypt == "0":
        # Exit the program
        print("\nEncrytion/Decription Program End.")
        break
        
    else:
        # Handle invalid main menu input
        print("Invalid Option. Please Enter E, D, or 0.\n")
        continue
```
