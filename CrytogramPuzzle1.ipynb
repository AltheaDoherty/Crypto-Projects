```python
#importing libraries
import tkinter as tk

#creating a pop out window for program
window = tk.Tk()
window.geometry("750x550") #window size
window.title("Cryptogram Puzzle") #window title

#creating health bar
health = tk.IntVar(value=100) #health bar status
hLabel=tk.Label(window,text="Health: 100") #health bar label
hLabel.pack(pady=10) #spacing

#creating label on top of where quote will sit
eLabel = tk.Label(window, text="Quote") 
eLabel.pack(pady=10)

#creating label on top of where entry window is
pLabel = tk.Label(window, text="Progress")
pLabel.pack(pady=10)
#size of the entry window
pFrame = tk.Frame(window)
pFrame.pack(pady=10)

#puzzle begin
encrypted = "IFMMP"
correct   = "HELLO"

#empty tuple to hold entries
entries = []

for i, ch in enumerate(encrypted):
    label = tk.Label(pFrame, text=ch, font=("Times",18))
    label.grid(row=0, column=i, padx=3)

    if ch.isalpha():
        entry = tk.Entry(pFrame, width=3, font=("Terminal",14), justify="center")
        entry.grid(row=1, column=i, padx=3)
        entries.append(entry)
    else:
        tk.Label(pFrame, text=" ").grid(row=1,column=i)

#fuction to update health bar
def hLvl(amount):
    hupdate= max(0, health.get()-amount)
    health.set(hupdate)
    hLabel.config(text= f"Health: {hupdate}")

#function to show correctness of guesses
def subGuess():
    guessl = (e.get().upper() for e in entries)
    wrong = False
    
    for i, guess in enumerate(guessl):
        if guess == correct[i]:
            entries[i].config(bg="limegreen")

        else:
            entries[i].config(bg="red")
            wrong = True

    if wrong:
        hLvl(10)

    for i, guess in enumerate(guessl):
        if guess and guess ==corect[i]:
            for x, actual in enumerate(correct):
                if actual ==guess:
                    entries[i].delete(0,tk.end)
                    entries[i].insert(0,guess)
                    entries[i].config(bg="limegreen")

#creating submit button to turn in guesses for puzzle
subButton = tk.Button(window, text="Submit Entry", command=subGuess)
subButton.pack(pady=10)

#calling main
window.mainloop()
```
